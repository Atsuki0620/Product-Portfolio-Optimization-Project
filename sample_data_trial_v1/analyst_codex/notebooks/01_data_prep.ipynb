# 01_data_prep
project_spec.md に基づき、販売・生産データを読み込み、集計テーブルと粗利マトリクスを生成するノート。

```python
from pathlib import Path
import sys
import pandas as pd

NOTEBOOK_ROOT = Path.cwd()
PROJECT_ROOT = NOTEBOOK_ROOT.parent
sys.path.append(str(PROJECT_ROOT / 'scripts'))
from data_pipeline import (
    ensure_directories,
    load_sales_data,
    load_production_data,
    summarize_sales,
    summarize_cost,
    build_margin_matrix,
    derive_segment_demand,
)

YEARS = [2022, 2023, 2024]
ensure_directories()
```

```python
sales_df = load_sales_data(YEARS)
production_df = load_production_data(YEARS)
display(sales_df.head())
display(production_df.head())
```

```python
sales_summary = summarize_sales(sales_df)
cost_summary = summarize_cost(production_df)
margin_matrix = build_margin_matrix(sales_summary, cost_summary)
segment_demand = derive_segment_demand(sales_summary)
margin_matrix.head()
```

```python
output_dir = PROJECT_ROOT / 'data' / 'intermediate'
output_dir.mkdir(parents=True, exist_ok=True)
sales_summary.to_parquet(output_dir / 'sales_summary.parquet', index=False)
cost_summary.to_parquet(output_dir / 'cost_summary.parquet', index=False)
margin_matrix.to_parquet(output_dir / 'margin_matrix.parquet', index=False)
segment_demand.to_parquet(output_dir / 'segment_demand.parquet', index=False)
print('中間ファイルを書き出しました。')
```
