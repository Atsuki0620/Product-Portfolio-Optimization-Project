# 04_reporting\n英語ラベルのグラフを生成し、1ページレポートに貼り込む画像を出力する。

```python
from pathlib import Path\nimport pandas as pd\nimport matplotlib.pyplot as plt\nplt.rcParams['font.family'] = 'DejaVu Sans'\nDATA_DIR = Path('..') / 'data' / 'intermediate'\n
```

```python
segment = pd.read_csv(DATA_DIR / 'allocation_segment_summary.csv')\nscenario = pd.read_csv(DATA_DIR / 'scenario_results.csv')\nfig, ax = plt.subplots(figsize=(6, 3))\nsegment.set_index('segment')['share'].plot(kind='bar', ax=ax, color='#4C72B0')\nax.set_title('Optimized Segment Mix')\nax.set_ylabel('Share')\nax.set_xlabel('Segment')\nfig.tight_layout()\nfig.savefig('../reports/img_segment_mix.png', dpi=150)\nplt.close(fig)\nfig, ax = plt.subplots(figsize=(6, 3))\nscenario.plot(x='scenario', y='total_margin', kind='bar', ax=ax, legend=False, color='#55A868')\nax.set_title('Scenario Impact on Total Margin')\nax.set_ylabel('Total Margin (JPY)')\nax.set_xlabel('Scenario')\nfig.tight_layout()\nfig.savefig('../reports/img_scenario_margin.png', dpi=150)\nplt.close(fig)\n
```

```python
import matplotlib.patches as patches
fig, ax = plt.subplots(figsize=(7, 2.5))
ax.axis('off')
steps = ['Data Ingest', 'Aggregation', 'Cost & Margin Calc', 'Greedy Allocation', 'Validation & Review']
x = 0.05
for step in steps:
    rect = patches.FancyBboxPatch((x, 0.4), 0.16, 0.3, boxstyle='round,pad=0.02', facecolor='#E0E0E0')
    ax.add_patch(rect)
    ax.text(x + 0.08, 0.55, step, ha='center', va='center', fontsize=9)
    x += 0.18
for i in range(len(steps) - 1):
    ax.annotate('', xy=(0.21 + 0.18 * i, 0.55), xytext=(0.18 * i + 0.13, 0.55), arrowprops=dict(arrowstyle='->'))
fig.tight_layout()
fig.savefig('../reports/img_flowchart.png', dpi=150)
plt.close(fig)
```
